```python
import random

class Student:
    def __init__(self, student_id, name, age, major, gpa, attendance, status):
        self.student_id = student_id
        self.name = name
        self.age = age
        self.major = major
        self.gpa = gpa
        self.attendance = attendance
        self.status = status

    def __str__(self):
        return (f"{self.student_id:<8} | {self.name:<20} | {self.age:<4} | "
                f"{self.major:<18} | {self.gpa:<5.2f} | {self.attendance:<6}% | {self.status}")


class StudentRecordSystem:
    def __init__(self):
        self.records = {}
        self._load_sample_data()

    def _load_sample_data(self):
        """Populates the system with a large set of realistic student records."""
        sample_students = [
            ("STU1001", "Aarav Sharma", 20, "Computer Science", 3.85, 94, "Active"),
            ("STU1002", "Priya Patel", 21, "Data Science", 3.92, 98, "Active"),
            ("STU1003", "Liam O'Connor", 19, "Mechanical Eng.", 3.10, 82, "Active"),
            ("STU1004", "Sophia Chen", 22, "Computer Science", 3.98, 99, "Dean's List"),
            ("STU1005", "Marcus Vance", 20, "Business Admin", 2.65, 74, "Probation"),
            ("STU1006", "Ananya Verma", 21, "Electrical Eng.", 3.74, 91, "Active"),
            ("STU1007", "Elena Rostova", 23, "Biotechnology", 3.88, 95, "Dean's List"),
            ("STU1008", "David Kim", 19, "Economics", 3.40, 88, "Active"),
            ("STU1009", "Rohan Gupta", 22, "Data Science", 3.55, 89, "Active"),
            ("STU1010", "Isabella Rossi", 20, "Psychology", 3.79, 93, "Active"),
            ("STU1011", "Noah Williams", 21, "Mechanical Eng.", 2.45, 68, "Probation"),
            ("STU1012", "Zoya Khan", 19, "Computer Science", 3.91, 96, "Dean's List"),
            ("STU1013", "Ethan Brooks", 22, "Business Admin", 3.22, 85, "Active"),
            ("STU1014", "Mei Lin", 20, "Electrical Eng.", 3.68, 92, "Active"),
            ("STU1015", "Kabir Mehta", 21, "Economics", 3.15, 80, "Active"),
            ("STU1016", "Chloe Dubois", 23, "Biotechnology", 3.95, 97, "Dean's List"),
            ("STU1017", "Lucas Silva", 20, "Data Science", 2.89, 79, "Active"),
            ("STU1018", "Neha Joshi", 19, "Psychology", 3.60, 90, "Active"),
            ("STU1019", "Daniel Miller", 22, "Computer Science", 3.45, 87, "Active"),
            ("STU1020", "Aisha Siddiq ui", 21, "Electrical Eng.", 3.82, 94, "Active"),
            ("STU1021", "Vikram Malhotra", 20, "Mechanical Eng.", 3.30, 86, "Active"),
            ("STU1022", "Hannah Abbott", 22, "Business Admin", 3.71, 91, "Active"),
            ("STU1023", "Arjun Nair", 19, "Data Science", 3.89, 95, "Dean's List"),
            ("STU1024", "Sofia Morales", 21, "Biotechnology", 2.75, 76, "Probation"),
            ("STU1025", "Ryan Cooper", 23, "Economics", 3.50, 88, "Graduating")
        ]
        for s in sample_students:
            self.add_student(Student(*s), silent=True)

    def add_student(self, student, silent=False):
        if student.student_id in self.records:
            if not silent:
                print(f"Error: Student ID {student.student_id} already exists.")
            return False
        self.records[student.student_id] = student
        if not silent:
            print(f"Student {student.name} ({student.student_id}) added successfully.")
        return True

    def get_student(self, student_id):
        return self.records.get(student_id.upper())

    def update_gpa(self, student_id, new_gpa):
        student = self.get_student(student_id)
        if not student:
            print("Student not found.")
            return
        student.gpa = new_gpa
        if new_gpa >= 3.85:
            student.status = "Dean's List"
        elif new_gpa < 2.75:
            student.status = "Probation"
        else:
            student.status = "Active"
        print(f"Updated {student.name}'s GPA to {new_gpa:.2f} (Status: {student.status}).")

    def delete_student(self, student_id):
        student_id = student_id.upper()
        if student_id in self.records:
            removed = self.records.pop(student_id)
            print(f"Deleted record for {removed.name} ({student_id}).")
        else:
            print("Student ID not found.")

    def display_all(self, filter_major=None):
        print("\n" + "=" * 88)
        print(f"{'ID':<8} | {'Name':<20} | {'Age':<4} | {'Major':<18} | {'GPA':<5} | {'Attend':<7} | {'Status'}")
        print("-" * 88)
        count = 0
        for student in self.records.values():
            if filter_major and filter_major.lower() not in student.major.lower():
                continue
            print(student)
            count += 1
        print("=" * 88)
        print(f"Total records displayed: {count}\n")

    def generate_analytics(self):
        if not self.records:
            print("No records available.")
            return
        total = len(self.records)
        avg_gpa = sum(s.gpa for s in self.records.values()) / total
        avg_att = sum(s.attendance for s in self.records.values()) / total
        top_student = max(self.records.values(), key=lambda s: s.gpa)
        probation_count = sum(1 for s in self.records.values() if s.status == "Probation")

        print("\n--- SYSTEM ANALYTICS ---")
        print(f"Total Enrolled Students : {total}")
        print(f"Average University GPA  : {avg_gpa:.2f}")
        print(f"Average Attendance      : {avg_att:.1f}%")
        print(f"Top Performer           : {top_student.name} ({top_student.major} - {top_student.gpa} GPA)")
        print(f"Students on Probation   : {probation_count}")
        print("------------------------\n")


def main():
    system = StudentRecordSystem()

    while True:
        print("=== STUDENT RECORD MANAGEMENT SYSTEM ===")
        print("1. View All Student Records")
        print("2. Filter Records by Major")
        print("3. Search Student by ID")
        print("4. Add New Student")
        print("5. Update Student GPA")
        print("6. Delete Student Record")
        print("7. View System Analytics")
        print("8. Exit")

        choice = input("\nSelect an option (1-8): ").strip()

        if choice == "1":
            system.display_all()
        elif choice == "2":
            major = input("Enter major to filter by (e.g., Computer Science, Data Science): ")
            system.display_all(filter_major=major)
        elif choice == "3":
            sid = input("Enter Student ID (e.g., STU1004): ")
            student = system.get_student(sid)
            if student:
                print("\nRecord Found:")
                print(student, "\n")
            else:
                print("\nStudent not found.\n")
        elif choice == "4":
            sid = input("Enter new ID (e.g., STU1026): ").upper()
            name = input("Enter Full Name: ")
            age = int(input("Enter Age: "))
            major = input("Enter Major: ")
            gpa = float(input("Enter GPA (0.0 - 4.0): "))
            att = int(input("Enter Attendance %: "))
            status = "Dean's List" if gpa >= 3.85 else ("Probation" if gpa < 2.75 else "Active")
            system.add_student(Student(sid, name, age, major, gpa, att, status))
        elif choice == "5":
            sid = input("Enter Student ID to update: ")
            gpa = float(input("Enter new GPA: "))
            system.update_gpa(sid, gpa)
        elif choice == "6":
            sid = input("Enter Student ID to delete: ")
            system.delete_student(sid)
        elif choice == "7":
            system.generate_analytics()
        elif choice == "8":
            print("Exiting Student Record System.")
            break
        else:
            print("Invalid selection. Please choose between 1 and 8.\n")


if __name__ == "__main__":
    main()
```
